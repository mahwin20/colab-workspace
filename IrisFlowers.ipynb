spliting the data into training and testing

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=123)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
x_train
```

building the model_decision tree

```python
model = DecisionTreeClassifier(criterion='entropy', random_state=1234)
```

for iris we are using decision tree model (there are many other models)

```python
model
```

```python
model_train = model.fit(x_train, y_train)
model_train
```

testing the model with testing Data
the model will take random elements so,every time you run new values will appear different

```python
y_pred = model_train.predict(x_test)
y_pred
```

```python
y_pred[0:10]
```

```python
y_test[:10]
```

Evaluating report5 baseed on y_test and y_pred

```python
import sklearn.metrics as metrics
confusion = metrics.confusion_matrix(y_test, y_pred)
confusion
```

accuracy based on y_test and y_pred

```python
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
class_wise = metrics.classification_report(y_test, y_pred)
print(class_wise)
```

predicting the newv sample

```python
new_sample = [[5.1,3.5,1.4,0.2]]
prediction = model_train.predict(new_sample)
predicted_label = le.inverse_transform(prediction)
print("Predicted Label:", predicted_label[0])
```
