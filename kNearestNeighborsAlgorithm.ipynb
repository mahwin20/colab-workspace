DATA PREPROCESSING (CLEANING THE UNWANTED DATA)

```python
from sklearn import preprocessing
le = preprocessing.LabelEncoder()
le.fit(df.species)
df['species'] = le.transform(df.species)
print(df)
```

```python
print(df.species)
```

```python
x=df.iloc[:,0:4]
y=df.iloc[:,4]
```

```python
x
```

```python
y
```

Spliting the data into training and testing

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

Builing the model k-Nearest Neighbors algorithm

```python
model = KNeighborsClassifier(n_neighbors=5)
```

```python
model
```

traing the model with traing data

```python
model_train = model.fit(x_train,y_train)
```

```python
model_train
```

test the model with testing Data



```python
y_pred = model_train.predict(x_test)
```

```python
y_pred
```

```python
y_pred[0:10]
```

```python
print(y_test[:10])
```

Evaluating the performane of the Model

```python
i=0
print("/n--------------------------------------------------------------------------")
print('%-25s %-25s %-25s' % ('Original Label', 'Predicted Label', 'Correct/Wrong'))
print("/n---------------------------------------------------------------------------")
for label in y_test[:10]:
  print('%-25s %-25s' % (label, y_pred[i]), end="")
  if (label == y_pred[i]):
    print('%-25s' % ('Correct'))
  else:
    print('%-25s' % ('Wrong'))
  i=i+1
  print("-----------------------------------------------------------------------------")
```

Confusion Matrix

```python
import sklearn.metrics as metrics
confusion =confusion_matrix(y_test, y_pred)
confusion
```

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
accuracy
```

```python
print('Accuracy of the Classifier is %0.2f' % accuracy_score(y_test, y_pred))
print("----------------------------------------------------------------------")
```

```python
print("\nClassifion report:\n")
class_wise=metrics.classification_report(y_test,y_pred)
print(class_wise )
```

predicting the new samples

```python
new_sample = [[5.1,3.5,1.4,0.2]]
prediction = model.predict(new_sample)
predicted_label = le.inverse_transform(prediction)
print("Predicted Label:", predicted_label)
```
